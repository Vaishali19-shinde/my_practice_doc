```python
#2.write a function to find the largest of three numbers

def largest_of_three(a,b,c):
  if a>=b and a>=c:
      return a
  elif b>=a and b>=c:
      return b
  else:
       return c
print(largest_of_three(10,34,22))
```

```python
#3.Write  a function to calculate Factorial
def factorial(num):
  fact=1
  while  num>0:
    fact=fact*num
    num=num-1
  return fact
print(factorial(5))
```

```python
#4.write a function to check prime number
def is_prime(num):
    if num<=1:
        return False
    i=2
    while i<num:
      if num%i==0:
        return False
      i=i+1
    return True
print(is_prime(19))
```

```python
#6.Write a function to count ovwels in a string

def count_vowel(text):
  vowels='aeiouAEIOU'
  count=0
  for ch in text:
     if ch in vowels:
         count=count+1
  return count
print(count_vowel("Vaishali Shinde"))
```

```python
#7.Write a function to check palindrome strin

def isplaindrome(text):
   cleaned= text.lower()
   return cleaned ==cleaned[::-1]
print(isplaindrome("madam"))
```

```python
#8. write a function to genrate fibonacci series

def fibonacci(n):
  series=[]
  a,b=0,1
  count=0
  while count<n:
    series.append(a)
    a,b=b,a+b
    count=count+1
  return series
print(fibonacci(8))
```

```python
#9.wrirte a function to remove duplicate from list
def remove_duplicate(items):
  result=[]
  for item in items:
    if item not in result:
        result.append(item)
  return result
print(remove_duplicate([1,2,3,3,21]))
```

```python
#10.write a function to sum any number of argument

def sum_all(*numbers):
  total=0
  for num in numbers:
    total=total+num
  return total
print(sum_all(1,8,5,7))
```

```python
#12.Write a functin to use Another Function as input
def square(num):
  return num *num
def apply_function(numbers,operation):
  result=[]
  for num in numbers:
    result.append(operation(num))
  return result
print(apply_function([1,2,3,4],square))
```
