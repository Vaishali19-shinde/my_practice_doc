```python
#Series tuple
tuple1=(12,33,44,55)
s=pd.Series(tuple1)
s
```

```python
#DataFrame tuple
tuple1=[('vaishali',29,'mumbai'),('varsha',22,'pune')]
df=pd.DataFrame(tuple1,columns=['name','age','city'])
df
```

```python
#DataFrame list[list]

data=[['vaishali',29,'mumbai'],
      ['varsha',22,'pune'],
       ['pooja',12,'sangli']]
data
```

```python
df=pd.DataFrame(data,columns=('Name','Age','City'))
df
```

```python
#Series list[list]

list1=[12,33,22,11]
s=pd.DataFrame(list1)
s
```

```python
#DataFrame list[tuple]
# Sample list of tuples

data_tuples = [
    ('varsha', 25, 'pune'),
    ('vaishali', 30, 'mumbai'),
    ('pooja', 35, 'sangli')
]
df = pd.DataFrame(data_tuples, columns=['Name', 'Age', 'City'])
df

```

```python
#Series list[tuple]
data_tuples = [
    ('varsha'),
    ('vaishali'),
    ('pooja')]
s=pd.Series(data_tuples)
s
```

```python
#Series with set
set1={"apple","banana","orange"}
s=pd.Series(list(set1))
s

```

```python
#DataFrame with set
data = {
    'Category': ['A', 'B'],
    'Tags': [{'promo', 'active'}, {'expired', 'archived'}]
}

df = pd.DataFrame(data)
df
```

```python
#series with dictionay
data_dict = {'Apple': 100, 'Banana': 150, 'Cherry': 200}

# 2. Convert to a Pandas Series
s = pd.Series(data_dict)

s
```

```python
#DataFrame with dictionary
data = {
    'Name': ['vaishali', 'varshaa', 'pooja'],
    'Age': [25, 30, 35],
    'City': ['sangli', 'pune', 'sangli']
}

df = pd.DataFrame(data)
df
```
