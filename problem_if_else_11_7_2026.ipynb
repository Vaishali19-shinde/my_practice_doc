```python
#16.write python program to input basic python salary of an employee and claculates it gross salary according its following
#basic salary <=10000 :HRA=20% ,DA =80%
#basic salary <=20000 : HRA=30%, DA=90%
#basic salary >20000 : HRA=35%, DA=95%

basic_salary = float(input("Enter the Basic Salary of the employee: "))

# Calculate HRA and DA based on basic salary slabs
if basic_salary <= 10000:
    hra = basic_salary * 0.20  # 20% HRA
    da = basic_salary * 0.80   # 80% DA

elif basic_salary <= 20000:
    hra = basic_salary * 0.30  # 30% HRA
    da = basic_salary * 0.90   # 90% DA

else:
    hra = basic_salary * 0.35  # 35% HRA
    da = basic_salary * 0.95   # 95% DA

# Calculate Gross Salary
gross_salary = basic_salary + hra + da

# Display the salary breakdown
print("\n--- Salary Breakdown ---")
print("Basic Salary : ₹", basic_salary)
print("HRA Allowance: ₹", hra)
print("DA Allowance : ₹", da)
print("------------------------")
print("GROSS SALARY : ₹",gross_salary)


```

```python
#17.write python program to input electricity until charge and calculate total electricity bill according to the given condition
#for first 50 units rs .0.50/unit
#for next 100 units rs .0.75/unit
#for next 100 units rs .0.1.25/unit
#for unit above 250 rs 1.50/unit
#An additional surcharges of 17%  is added to the bill

units = float(input("Enter total electricity units consumed: "))

base_bill = 0.0

# Calculate base charge according to your specific tiered slabs
if units <= 50:
    base_bill = units * 0.50

elif units <= 150:
    # First 50 units at 0.50, next units up to 150 at 0.75
    base_bill = (50 * 0.50) + ((units - 50) * 0.75)

elif units <= 250:
    # First 50 at 0.50, next 100 at 0.75, next units up to 250 at 1.25
    base_bill = (50 * 0.50) + (100 * 0.75) + ((units - 150) * 1.25)

else:
    # First 50 at 0.50, next 100 at 0.75, next 100 at 1.25, units above 250 at 1.50
    base_bill = (50 * 0.50) + (100 * 0.75) + (100 * 1.25) + ((units - 250) * 1.50)

# Calculate the additional 17% surcharge
surcharge = base_bill * 0.17
total_bill = base_bill + surcharge

# Display the final bill breakdown
print("\n--- Electricity Bill Breakdown ---")
print("Units Consumed : ",units, " units")
print("Base Bill      : Rs. ", base_bill)
print("Surcharge (17%): Rs. ",surcharge)
print("----------------------------------")
print("TOTAL BILL     : Rs. ",total_bill)

```
