```python
#4.passing one function to output to Another Function
def square(num):
  return num*num
def double(num):
  return num*2
result=double(square(5))
print("final result",result)
```

```python
#7.variable number of Argument using *args
def add_all(*number):
  total=0
  for n in number:
    total = total + n
  return total
print("Total",add_all(10,20,30,29))
```

```python
#8.variable number of keyword Argument using **kwargs
def print_profile(**details):
  for key,value in details.items():
    print(key,":",value)
print_profile(name="vaishali",course="python",city="Sangli")
```

```python
#9 Local and Global VAriable
message= "global variable"
def show_message():
  message="local message"
  print("inside function:",message)
show_message()
print("outside function",message)
```

```python
#10.Lambda function
square = lambda n:n*n
print("square",square(6))
```

```python
#11.Recursive function
def factorial(n):
  if n==0:
    return 1
  return n* factorial(n-1)
print("Factorial",factorial(5))
```

```python
#12.Function with Docdtring and Type Hints
def simple_interest(principle:float,rate:float,time:float)->float:
  return (principle * rate*time)/100
intrerest=simple_interest(10000,8,2)
print("simple interest:",intrerest)
```
