```python
#1.reverse a number

no=int(input("Enter a number"))
rev=0
while (no > 0):
   digit=no%10
   rev=rev*10+digit
   no=no//10
print("reverse number",rev)
```

```python
#2.check number is palindrome or not
no=int(input("Enter a number=:"))
temp=no
rev=0
while(no>0):
  digit=no%10
  rev=rev*10+digit
  no=no//10
if temp==rev:
  print("number is palindrome")
else:
  print("number is not palindrome")
```

```python
#3.count digit in a number
number=int(input("Enter a number="))
count=0
while(number>0):
    count=count + 1
    number=number//10
print("number count is=", count)
```

```python
#4.Find sum of digit

number=int(input("Enter a number:="))
sum=0
while(number>0):
  digit=number%10
  sum=sum+digit
  number=number//10
print("sum of digit==",sum)
```

```python
#5.check whether number is armstrong or not
number=int(input("Enter a number"))
temp=number
rev=0
while(number>0):
  digit=number%10
  rev=rev+digit**3
  number=number//10
if temp==rev:
  print("Number is armstrong:=",temp)
else:
  print("Number is not Armstrong:=",temp)
```

```python
#6.check weather number is prime
number=int(input("Enter a number"))
i=2
is_prime=True
while i<number:
  if number % i==0:
      is_prime = False
      break
  i=i+1
if is_prime and number>1:
  print("prime number")
else:
  print("not prime number")

```

```python
#7.Find factorial of a number
num=int(input("Entera number="))
fact=1
while num>0:
    fact=fact*num
    num=num-1
print("Factorial",fact)
```

```python
#8.Genrate Fibonassi series
n=int(input("Enter anumber:="))
a=0
b=1
count=0
while count<n:
    print(a)
    c=a+b
    a=b
    b=c
    count=count+1
```
