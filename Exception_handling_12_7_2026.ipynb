```python
#1.why Exception handling is needed
a=10
b=0
try:
    result=a/b
    print(result)
except ZeroDivisionError:
    print("Can not divide by 0")
```

```python
#2.Basic try except block
try:
    age=int(input("Enter your age="))
    print("Your age is=",age)
except ValueError:
    print("please enter valid number")
```

```python
#3.Handling multiple exception
try:
    a=int(input("Entera a number"))
    b=int(input("Enter b number"))
    print("Result",a/b)
except ValueError:
    print("Invalid Input"",""please enter a valid number")
except ZeroDivisionError:
    print("Second number can not be Zero")
```

```python
#4.using except Exception as e
try:
   numbers=[12,20,30]
   print(numbers[5])
except Exception as e:
   print("Error Occured:",e)

```

```python
#5.using else with try-except
try:
  num=int(input("enter a number:="))
except ValueError:
  print("Invalid Number:=")
else:
  print("You entered number:=",num)
```

```python
#6.using finally block
try:
    file=open("example.txt","r")
    print(file.read())
except FileNotFoundError:
    print("File not found")
finally:
   print("Program executed completed")
```

```python
#7.Raising an Exception Manually
age=16
try:
  if age<18:
    raise ValueError("Age must be 18 or above")
    print("Eligible")
except ValueError as e:
    print(e)
```

```python
#8.Create a custon exception
class insufficentbalanceError(Exception):
    pass
balance=500
withdraw=1000
try:
  if withdraw>balance:
    raise insufficentbalanceError("insufficient account balance")
    balnce=balance-withdraw
    print("Withdraw successful")
except insufficentbalanceError as e:
    print(e)



```

```python
#9.Exception hndling in file reading
try:
  with open("data.txt","r") as file:
    content=file.read()
    print(content)
except FileNotFoundError:
    print("The file does not exit")
```

```python
#10.Nested try except
try:
      num=int(input("enter a number"))
      try:
         result=100/num
         print("Result",result/100)
      except ZeroDivisionError:
        print("can not divided by zero")
except ValueError:
   print("Invalid input")
```

```python
#11.Exception handling with list access
items=["python","java","html"]
try:
   print(items[4])
except IndexError:
   print("Invalid index selected")
```

```python
#12.Using try except using function
def safe_divide(a,b):
    try:
         return a/b
    except ZeroDivisionError:
         return "can not divided by zero"
print(safe_divide(10,2))
print(safe_divide(10,0))
```
